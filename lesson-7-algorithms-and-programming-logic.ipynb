1. Crie uma matriz 3x3 com valores digitados pelo usuário.

```python
linhas = 3
colunas = 3
matriz = []

for i in range(linhas):
  linha = []
  for j in range(colunas):
    valor = int(input(f"Digite o valor para [{i}] e [{j}]: "))
    linha.append(valor)
  matriz.append(linha)

print("Matriz  resultante:")
for linha in matriz:
  print(linha)
```

2. Some todos os elementos de uma matriz 3x3.

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

soma = 0

for linha in matriz:
  for numero in linha:
    soma += numero

print(soma)
```

5. Conte quantos números pares existem em uma matriz.

```python
matriz = [
    [1, 2],
    [3, 4]
]

contador = 0

for linha in matriz:
  for numero in linha:
    if numero % 2 == 0:
      contador += 1

print(contador)
```

6. Calcule a média de todos os elementos de uma matriz.

```python
matriz = [
    [1, 2],
    [3, 4]
]

soma = 0
contador = 0

for linha in matriz:
  for numero in linha:
    soma += numero
    contador += 1

media = soma / contador

print(media)
```

11. Verifique se duas matrizes 2x2 digitadas pelo usuário são iguais.

```python
linhas1 = 2
colunas1 = 2
matriz1 = []

linhas2 = 2
colunas2 = 2
matriz2 = []

for i in range(linhas1):
  linha1 = []
  for j in range(colunas1):
    valor = int(input(f"Digite o valor para [{i}] e [{j}]: "))
    linha1.append(valor)
  matriz1.append(linha1)

print("Matriz 1 resultante:")
for linha1 in matriz1:
  print(linha1)

for i in range(linhas2):
  linha2 = []
  for j in range(colunas2):
    valor = int(input(f"Digite o valor para [{i}] e [{j}]: "))
    linha2.append(valor)
  matriz2.append(linha2)

print("Matriz 2 resultante:")
for linha2 in matriz2:
  print(linha2)

if linha1 == linha2:
  print("As matrizes são iguais")
else:
  print("As matrizes não são iguais")
```
