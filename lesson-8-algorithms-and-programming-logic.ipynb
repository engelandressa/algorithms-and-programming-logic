3. Crie uma função que receba uma lista de números e retorne quantos são pares.  

```python
def par(num):
  return num % 2 == 0

def contar(lista):
  return len([n for n in lista if par (n)])

lista = [1, 2, 3]

print(contar(lista))
```

4. Crie uma função chamada `eh_primo(numero)` que retorne `True` se o número for primo.  

```python
def primo(num):
  if num <=1:
    return False
  for i in range(2, num):
    if num % i == 0:
      return False
  return True

print(primo(3))
print(primo(1))
print(primo(5))
print(primo(10))
```

5. Crie uma função `converter_para_fahrenheit(celsius)` que converta graus Celsius para Fahrenheit.

```python
def converter_faherenheit(valor):
  while True:
    valor = float(valor)
    valor = (valor*1.8) + 32
    return (f"{valor:.0f}°F")

print(converter_faherenheit(10))
print(converter_faherenheit(24))
print(converter_faherenheit(30))
```

9. Crie uma função que conte o número de vogais em uma palavra.  

```python
def vogais(palavra):
  palavra = palavra.lower()
  contador = 0
  lista_vogais = ["a", "i", "u", "e", "o"]

  for letra in palavra:
    if letra in lista_vogais:
      contador += 1
  return contador

print(vogais("Jane"))
print(vogais("John"))
print(vogais("Andressa"))
```

10. Crie um menu com funções para:
    - Cadastrar nomes
    - Listar nomes
    - Sair do programa

```python
nomes = []

def cadastrar():
  nome = input("Digite o nome: ")
  nomes.append(nome)
  print("Nome cadastrado com sucesso!")

def listar():
  if len(nomes) == 0:
    print("Nenhum nome cadastrado.")
  else:
    print("Lista de nomes:")
    for n in nomes:
      print("-", n)


def sair():
  print("Saindo do programa...")

while True:
    print("\nMENU:")
    print("1 - Cadastrar nome")
    print("2 - Listar nomes")
    print("3 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        cadastrar()
    elif opcao == "2":
        listar()
    elif opcao == "3":
        sair()
        break
    else:
        print("Opção inválida. Tente novamente.")
```
