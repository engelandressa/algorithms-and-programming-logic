9. Crie uma lista com 5 notas de alunos, calcule a média e diga quais alunos ficaram acima da média.

```python
sam = [5, 5, 5]
bob = [10, 5, 10]
dan = [9, 8, 9]
ana = [8, 8, 8]
tom = [3, 5, 10]

alunos = [sam, bob, dan, ana, tom]
nomes = ["Sam", "Bob", "Dan", "Ana", "Tom"]

medias = []

for aluno in alunos:
    media = sum(aluno) / len(aluno)
    medias.append(media)

media_turma = sum(medias) / len(medias)

print("Média da turma:", media_turma)
print("Alunos acima da média:")

for i in range(len(medias)):
    if medias[i] > media_turma:
        print(nomes[i])
```

```python
sam = [5, 5, 5]
bob = [10, 5, 10]
dan = [9, 8, 9]
ana = [8, 8, 8]
tom = [3, 5, 10]

# Dicionário com nomes e notas
alunos = {
    "Sam": sam,
    "Bob": bob,
    "Dan": dan,
    "Ana": ana,
    "Tom": tom
}

# Calcula a média geral da turma
todas_as_notas = [nota for notas in alunos.values() for nota in notas]
media_geral = sum(todas_as_notas) / len(todas_as_notas)

print(f"Média da turma: {media_geral:.2f}\n")

# Verifica quem ficou acima da média
print("Alunos acima da média:")
for nome, notas in alunos.items():
    media = sum(notas) / len(notas)
    if media > media_geral:
        print(f"{nome} com média {media:.2f}")
```

17. Faça uma função que recebe uma lista de números e retorna uma nova lista com o fatorial de cada número.

```python
def fatorial(n):
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado

def lista_fatoriais(lista):
    fatlist = []
    for num in lista:
        fatlist.append(fatorial(num))
    return fatlist

numlist = []

for i in range(3):
    num = int(input("Digite um número: "))
    numlist.append(num)

resultado = lista_fatoriais(numlist)
print(resultado)
```

:18. Faça uma função que recebe uma lista e retorna outra com os elementos em ordem reversa (sem usar `.reverse()` ou `[::-1]`).

```python
words = []

for i in range(0,3):
  word = input(f"Write the word: ")
  words.append(word)

print(words)

def list_invert(lista):
    list_invert = []

    for i in range(len(lista) - 1, -1, -1):
        list_invert.append(lista[i])

    return list_invert


result = list_invert(words)
print(result)
```

19. Simule um carrinho de compras: adicione produtos até que o usuário digite 'fim' e, no final, mostre o carrinho.

```python
compras = []

print("Adicione os produtos (digite 'fim' para encerrar)")

while True:
    produto = input("Digite o nome do produto: ")

    if produto.lower() == "fim":
        break
    else:
        compras.append(produto)

print("Carrinho de compras:")
for item in compras:
    print("-", item)
```

20. Desenvolva um menu de opções para gerenciar uma lista de tarefas: adicionar, remover, exibir e sair.

```python
tarefas = []

while True:
    print("\nMENU DE TAREFAS")
    print("1 - Adicionar tarefa")
    print("2 - Remover tarefa")
    print("3 - Exibir tarefas")
    print("4 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        tarefa = input("Digite a tarefa: ")
        tarefas.append(tarefa)
        print("Tarefa adicionada!")

    elif opcao == "2":
        if len(tarefas) == 0:
            print("Não há tarefas para remover.")
        else:
            for i in range(len(tarefas)):
                print(f"{i} - {tarefas[i]}")

            indice = int(input("Digite o número da tarefa para remover: "))

            if 0 <= indice < len(tarefas):
                tarefas.pop(indice)
                print("Tarefa removida!")
            else:
                print("Índice inválido.")

    elif opcao == "3":
        if len(tarefas) == 0:
            print("Lista de tarefas vazia.")
        else:
            print("Tarefas:")
            for i, tarefa in enumerate(tarefas):
                print(f"{i} - {tarefa}")

    elif opcao == "4":
        print("Saindo do programa...")
        break

    else:
        print("Opção inválida.")
```
