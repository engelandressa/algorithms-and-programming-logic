7. Faça um programa que leia 10 números e armazene em duas listas: uma com pares e outra com ímpares.

```python
numerospar = []
numerosimpar = []

for i in range(0,10):
  numero = int(input(f"Digite o {i+1}º número: "))
  if numero % 2 == 0:
    numerospar.append(numero)
  elif numero % 2 != 0:
    numerosimpar.append(numero)
print(f"Lista do números pares: {numerospar}")
print(f"Lista do números ímpares: {numerosimpar}")
```

```python
sam = [5, 5, 5]
bob = [10, 5, 10]
dan = [9, 8, 9]
ana = [8, 8, 8]
tom = [3, 5, 10]

# Dicionário com nomes e notas
alunos = {
    "Sam": sam,
    "Bob": bob,
    "Dan": dan,
    "Ana": ana,
    "Tom": tom
}

# Calcula a média geral da turma
todas_as_notas = [nota for notas in alunos.values() for nota in notas]
media_geral = sum(todas_as_notas) / len(todas_as_notas)

print(f"Média da turma: {media_geral:.2f}\n")

# Verifica quem ficou acima da média
print("Alunos acima da média:")
for nome, notas in alunos.items():
    media = sum(notas) / len(notas)
    if media > media_geral:
        print(f"{nome} com média {media:.2f}")
```

10. Faça um programa que leia números do usuário até que ele digite 0. Depois, mostre a lista e a soma dos números.

```python
numeros = []
soma = 0

print("Digite os números e 0 para parar")
while True:
  numero = int(input(f"Digite o numero: "))
  if numero != 0:
    numeros.append(numero)
    soma += numero
  elif numero == 0:
    break
print(f"Os números digitados foram: {numeros}")
print(f"A soma dos números é {soma}")
```

12. Substitua todos os números ímpares de uma lista por zero.

```python
numeros = [0,-9, 2, -4, 19, 3 ,-9]
impar = [0 if n % 2 != 0 else n for n in numeros]
print(impar)
```

13. Leia uma lista de números e crie uma nova lista apenas com os valores únicos (sem repetições).
